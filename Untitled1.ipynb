```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
```

```python
from sklearn.linear_model import Perceptron # Used for simple linear classification task

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

from tensorflow.keras.models import Sequential # Sequential lets you build  a neural network layer-by-layer in keras

from tensorflow.keras.layers import Dense # Dense make the final prediction
from tensorflow.keras.layers import Conv2D # Conv2D extract the features
from tensorflow.keras.layers import Flatten # Flatten reshape them

from tensorflow.keras.layers import MaxPooling2D # Maxpooling2D reduce the size
from tensorflow.keras.layers import Dropout # Dropout prevent

from tensorflow.keras.utils import to_categorical # convert numeric class label into one-hot encoded format for training classification models
```

```python
df_train = pd.read_csv('mnist_train.csv')
df_test = pd.read_csv('mnist_test.csv')
```

```python
# Preprocess
X_train = df_train.drop('label', axis=1).values
y_train = df_train['label'].values

X_test = df_test.drop('label', axis=1).values
y_test = df_test['label'].values
```

```python
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0
```

```python
X_train_img = X_train.reshape(-1, 28, 28)
X_test_img = X_test.reshape(-1, 28, 28)
```

```python
X_train_img.shape
```

```python
X_train.shape
```

```python
df_train['label'].value_counts() # checked how many unique number in label and how much times use that number
```

```python
y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)
```

```python
y_train_cat.shape
```

```python
perceptron = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(10, activation='softmax')
])
```

```python
perceptron.compile(optimizer='sgd', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
history_recp = perceptron.fit(X_train_img, y_train_cat, epochs=10, batch_size=32, validation_data=(X_test_img, y_test_cat), verbose=1)
```

```python
y_train_cat.shape
```

```python
X_train_img.shape
```

```python
print("X_train_img:", X_train_img.shape)
print("y_train_cat:", y_train_cat.shape)
print("X_test_img:", X_test_img.shape)
print("y_test_cat:", y_test_cat.shape)
```

```python
acc_percp = perceptron.evaluate(X_test_img, y_test_cat, verbose=1)[1]
```

```python
ann = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation='relu'),
    Dense(64, activation='relu'),
    Dense(32, activation='relu'),
    Dense(16, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
ann.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
ann.fit(X_train_img, y_train_cat, epochs=10, batch_size=32, validation_data=(X_test_img, y_test_cat), verbose=1)
```

```python
acc_ann = ann.evaluate(X_test_img, y_test_cat, verbose=1)[1]
```

```python
X_train_cnn = X_train.reshape(-1, 28, 28, 1)
X_test_cnn = X_test.reshape(-1, 28, 28, 1)
```

```python
cnn = Sequential([
    Conv2D(128, kernel_size=(3,3), activation='relu', input_shape=(28, 28, 1)),
    MaxPooling2D(pool_size=(2,2)),
    Conv2D(64, kernel_size=(3,3), activation='relu'),
    MaxPooling2D(pool_size=(2,2)),
    Conv2D(32, kernel_size=(3,3), activation='relu'),
    MaxPooling2D(pool_size=(2,2)),
    Flatten(),
    Dropout(0.5),
    Dense(10, activation='softmax')
])
```

```python
cnn.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
history_cnn = cnn.fit(X_train_cnn, y_train_cat, epochs=10, batch_size=32, validation_data=(X_test_cnn, y_test_cat), verbose=1)
```

```python
acc_cnn = cnn.evaluate(X_test_cnn, y_test_cat, verbose=1)[1]
```
